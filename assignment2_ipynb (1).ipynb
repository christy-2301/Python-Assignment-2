***List Creation:***

```python
"""1.a"""

age_list = [24, 25, 26, 27, 28]                                        # Create a list of ages
name_list = ["christy", "aishu", "ardra", "kezii", "alwin"]            # Create a list of names
print(age_list)

"""1.b"""
print(name_list)
```

***List Operations / Modifications:***

```python
"""2.a"""

name_list.append("Yazhini")                                        # Add a name to the list
print(name_list)
```

```python
"""2.b"""

age_list.insert(2,30)                                           # Insert an age at index 2
print(age_list)
```

```python
"""2.c"""

name_list.remove("Yazhini")                                    # Remove a name from the list
print(name_list)
```

```python
"""2.d"""

age_list.pop()                                               # Remove the last age
print(age_list)
```

```python
"""2.e"""

age_list.extend([29,30,26])                                 # Extend the list with multiple ages
print(age_list)
```

```python
"""2.f"""

age_list.sort(reverse=True)                                 #Sort the ages in descending order

print(age_list)
```

```python
"""2.g"""

print(max(age_list))                                        # Find the maximum age
print(min(age_list))                                        # Find the minimum age
print(sum(age_list))                                        # Find the total of ages
```

```python
"""3.a"""

print(name_list[0])                                         # Print the first name
```

```python
"""3.b"""

print(name_list[-1])                                        # Print the last name
```

```python
"""3.c"""

print(name_list[2:5])                                       # Print elements from index 2 to index 4
```

```python
"""3.d"""

name_list.reverse()                                         # Reverse the name list
print(name_list)
```

***Dictionary (Creation, Modification and Access):***

```python
"""a"""

student_marks ={"christy":95,                               # Create a dictionary of student marks
               "aishu":81,
               "ardra":75,
               "kezii":92,
               "alwin":85}
print(student_marks)
```

```python
"""b"""

print(student_marks["christy"])                                  # Print the mark of a student
```

```python
"""c"""

student_marks["Janani"]=80                                      # Add a new student
print(student_marks)
```

```python
"""d"""

student_marks["ardra"]=82                                      # Update a student's mark
print(student_marks)
```

```python
"""e"""

print(student_marks.keys())                                   # Print dictionary keys
print(student_marks.values())                                 # Print dictionary values
print(student_marks.items())                                  # Print dictionary items
```

***Sets (Operations):***

```python
"""a"""

my_set={'a','e','i','o','u','a','a','i'}                    # Create a set with vowels
print(my_set)
print(type(my_set))
```

```python
"""b"""

my_set[4] = 's'                                           # Try to modify a set using an index number

print(my_set)
```

```python
"""c"""

set1={1, 3, 5, 7, 9}                                            # Create the first set
set2={2, 3, 5, 8, 10}                                           # Create the second set
print(set1)
print(set2)
```

```python
"""d - Union"""

print(set1|set2)                                            # Find the union of two sets
```

```python
"""d - Intersection"""

print(set1 & set2)                                         # Find the intersection of two sets
```

***Operators & Conditional Statements :
(IF, ELIF, ELSE)***

```python
score=int(input("enter the score: "))                                      # Get the score from the user
if score < 0 or score > 10:                                                # Check if the score is valid
    print("INVALID SCORE \n That's not a valid score!❌❌❌")
elif score>7:                                                              # Check for above average
    print("ABOVE AVERAGE \n Excellent performance. Keep up the good work!👍🌟")
elif 4 <= score <= 7:                                                      # Check for average
    print("AVERAGE \n Good effort. Keep practicing to improve your performance.👍")
else :                                                                     # Check for below average
    print("BELOW AVERAGE \n Need to improve your performance. Consistent practice will lead to better results.📚 ")
```
